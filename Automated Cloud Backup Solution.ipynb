```python
import os
import shutil

with open("sample.txt", "w") as file:
    file.write("This is my Automated Cloud Backup Project.")


backup_folder = "CloudBackup"
os.makedirs(backup_folder, exist_ok=True)


shutil.copy("sample.txt", backup_folder)

print("Backup Successful!")

print("\nFiles in CloudBackup:")
print(os.listdir(backup_folder))
```
